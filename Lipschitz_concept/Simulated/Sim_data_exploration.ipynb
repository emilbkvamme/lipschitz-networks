# Data Exploration for the $q\bar{q}\rightarrow Zg$ simulated process

```python
import os

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# PATHS, for this notebook only. Jupyter runs a notebook from its own folder, so NOTEBOOK_DIR is that folder and the
# paths below work on any machine. Change them here to read from, or write to, somewhere else.
NOTEBOOK_DIR = os.path.abspath("")
DATA_PATH = os.path.join(NOTEBOOK_DIR, "zg.npy")   # the simulated qq->Zg events
```

```python
# Loading data
sim_data_path = DATA_PATH

file = np.load(sim_data_path, allow_pickle=True)
```

```python
# Dimensions
print("Dimensions of the data:", file.shape)

# Make distribution plots for each variable
for i in range(file.shape[1]):
    plt.figure(figsize=(8, 6))
    if np.all(file[:, i] > 0):  # Check if all values are positive
        plt.hist(np.log10(file[:, i]), bins=50, density=True, alpha=0.7, color='blue')
        plt.title(f'Distribution of Variable {i+1} (Log Scale)')
        plt.xlabel(f'Log10 of Variable {i+1} Values')
    else:
        plt.hist(file[:, i], bins=50, density=True, alpha=0.7, color='blue')
        plt.title(f'Distribution of Variable {i+1}')
        plt.xlabel(f'Variable {i+1} Values')
    plt.title(f'Distribution of Variable {i+1}')
    plt.xlabel(f'Variable {i+1} Values')
    plt.ylabel('Density')
    plt.yscale('log')  # Set y-axis to logarithmic scale
    plt.show()
```

```python
# Checking last variable
log_bins = np.logspace(np.log10(min(file[:, -1])), np.log10(max(file[:, -1])), num=50)
plt.figure(figsize=(8, 6))
plt.hist(file[:, -1], bins=log_bins, density=True, alpha=0.7, color='green')
plt.title('Distribution of Last Variable')
plt.xlabel('Last Variable Values')
plt.ylabel('Density')
plt.xscale('log')  # Set x-axis to logarithmic scale
#plt.yscale('log')  # Set y-axis to logarithmic scale
plt.show()
```

```python
# Checking last variable
#log_bins = np.logspace(np.log10(min(file[:, -1])), np.log10(max(file[:, -1])), num=50)
plt.figure(figsize=(8, 6))
plt.hist(np.log10(file[:, -1]), bins=50, density=True, alpha=0.7, color='green')
plt.title('Distribution of Last Variable')
plt.xlabel('Last Variable Values')
plt.ylabel('Density')
#plt.xscale('log')  # Set x-axis to logarithmic scale
#plt.yscale('log')  # Set y-axis to logarithmic scale
plt.show()
```

## Testing a standard scaler on the distributions

```python
from sklearn.preprocessing import StandardScaler
```

```python

scaler = StandardScaler()
scaled_data = scaler.fit_transform(file)
```

```python
# Plotting the scaled data distributions
for i in range(scaled_data.shape[1]):
    plt.figure(figsize=(8, 6))
    plt.hist(scaled_data[:, i], bins=50, density=True, alpha=0.7, color='orange')
    plt.title(f'Distribution of Scaled Variable {i+1}')
    plt.xlabel(f'Scaled Variable {i+1} Values')
    plt.ylabel('Density')
    plt.yscale('log')  # Set y-axis to logarithmic scale
    plt.show()
```

```python
# Checking last variable
plt.figure(figsize=(8, 6))
plt.hist(scaled_data[:, -1], bins=50, density=True, alpha=0.7, color='green')
plt.title('Distribution of Last Variable')
plt.xlabel('Last Variable Values')
plt.ylabel('Density')
#plt.xscale('log')  # Set x-axis to logarithmic scale
plt.yscale('log')  # Set y-axis to logarithmic scale
plt.show()
```
